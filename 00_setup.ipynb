```sql
create volume if not exists tnelections.bronze.form20;
```

```sql
create schema if not exists tnelections.form20_bronze;
create schema if not exists tnelections.form20_silver;
create schema if not exists tnelections.form20_gold;
```

```sql
SELECT table_catalog AS catalog, table_schema AS schema, table_name AS name, table_type AS type
FROM system.information_schema.tables
WHERE table_catalog IN ('tnelections', 'elections')
  AND table_schema NOT IN ('information_schema', 'form20_bronze', 'form20_silver', 'form20_gold')

UNION ALL

SELECT volume_catalog, volume_schema, volume_name, 'VOLUME'
FROM system.information_schema.volumes
WHERE volume_catalog IN ('tnelections', 'elections')
  AND volume_schema NOT IN ('information_schema', 'form20_bronze', 'form20_silver', 'form20_gold')

ORDER BY 1, 2, 3;
```
